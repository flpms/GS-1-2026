Esse notebook simula um sensor neuromórfico de baixo consumo para detectar uma condição crítica em um ambiente espacial.

**Cenário:** um rover na superfície de Marte, operando longe da estação central.

**O que ele detecta:** o aumento de radiação, temperatura e poeira ao mesmo tempo durante uma tempestade
marciana, que coloca em risco os equipamentos e a tripulação.

A ideia é usar três conceitos que vimos no semestre: memristor virtual (memória resistiva), limiar de
disparo e processamento por eventos. Em vez de mandar todo dado bruto para a Terra, o sensor acumula a
condição ali mesmo e só envia um aviso (NORMAL, OBSERVACAO ou ALERTA_CRITICO) quando precisa, o que
economiza energia e sinal.

## 3. Função do sensor neuromórfico

A função pega temperatura, radiação e poeira e junta tudo em uma tensão de entrada (`V_entrada`). Depois
ela simula um memristor virtual com memória: quando a tensão passa do limiar, o estado interno (`w`) sobe;
quando fica abaixo, ele desce devagar. O LED mostra o resultado.

```python
def simular_sensor(dados, V_limiar, taxa_chaveamento,
                   limiar_estado_alerta=0.65,
                   limiar_estado_observacao=0.35):
    saida = dados.copy()

    # Normalizacao simples das variaveis fisicas (0 a 1)
    temp_norm = np.clip((saida['temperatura_C'] - 20) / (55 - 20), 0, 1)
    rad_norm = np.clip((saida['radiacao_uSv_h'] - 0.25) / (2.5 - 0.25), 0, 1)
    poeira_norm = np.clip(saida['poeira_pct'] / 100, 0, 1)

    # Conversao conceitual para tensao de entrada do circuito
    indice_sensor = 0.45 * temp_norm + 0.35 * rad_norm + 0.20 * poeira_norm
    saida['V_entrada'] = np.round(15 + 25 * indice_sensor, 2)

    estado_memristor = []
    led = []
    diagnostico_sensor = []
    w = 0.0
    dt = 5  # intervalo entre leituras, em minutos

    for v in saida['V_entrada']:
        if v > V_limiar:
            # O memristor acumula efeito quando a tensao passa do limiar
            w = min(1.0, w + taxa_chaveamento * (v - V_limiar) * dt * (1 - w))
        else:
            # Pequeno relaxamento quando o sinal fica abaixo do limiar
            w = max(0.0, w - 0.002 * dt * w)

        estado_memristor.append(round(w, 3))

        if w >= limiar_estado_alerta:
            led.append('VERMELHO')
            diagnostico_sensor.append('ALERTA_CRITICO')
        elif w >= limiar_estado_observacao:
            led.append('AMARELO')
            diagnostico_sensor.append('OBSERVACAO')
        else:
            led.append('APAGADO')
            diagnostico_sensor.append('NORMAL')

    saida['estado_memristor'] = estado_memristor
    saida['LED'] = led
    saida['diagnostico_sensor'] = diagnostico_sensor
    return saida
```

## 4. Ajustes do grupo

Três ajustes diferentes do sensor, dentro das faixas do enunciado (V_limiar: 24 a 34 / taxa: 0.003 a 0.008).
O Ajuste A foi alterado em relação ao valor inicial sugerido (era 24 / 0.006), conforme exigido na atividade.

```python
ajustes_da_equipe = {
    'Ajuste_A_sensivel':    {'V_limiar': 26, 'taxa_chaveamento': 0.007},  # alterado pelo grupo
    'Ajuste_B_equilibrado': {'V_limiar': 28, 'taxa_chaveamento': 0.005},
    'Ajuste_C_conservador': {'V_limiar': 32, 'taxa_chaveamento': 0.004},
}

ajustes_da_equipe
```

## 5. Rodar a simulação

Executa os três ajustes e mostra em que minuto o sensor começou a indicar observação (amarelo)
e alerta crítico (vermelho).

```python
resultados = {}
resumo = []

for nome, pars in ajustes_da_equipe.items():
    saida = simular_sensor(df, pars['V_limiar'], pars['taxa_chaveamento'])
    resultados[nome] = saida

    primeiro_nao_apagado = saida[saida['LED'] != 'APAGADO'].head(1)
    primeiro_vermelho = saida[saida['LED'] == 'VERMELHO'].head(1)

    resumo.append({
        'ajuste': nome,
        'V_limiar': pars['V_limiar'],
        'taxa_chaveamento': pars['taxa_chaveamento'],
        'primeiro_LED_amarelo_min': None if primeiro_nao_apagado.empty else int(primeiro_nao_apagado['tempo_min'].iloc[0]),
        'primeiro_LED_vermelho_min': None if primeiro_vermelho.empty else int(primeiro_vermelho['tempo_min'].iloc[0]),
    })

resumo_transicoes = pd.DataFrame(resumo)
display(resumo_transicoes)
```

## 6. Visualização da memória do memristor

O gráfico mostra a memória do memristor (`w`) subindo nos três ajustes ao longo do tempo. As linhas
tracejadas marcam os limiares de observação (0.35) e de alerta crítico (0.65). Quanto mais sensível o
ajuste, mais cedo a curva cruza essas linhas.

```python
plt.figure(figsize=(9, 5))
for nome, saida in resultados.items():
    plt.plot(saida['tempo_min'], saida['estado_memristor'], label=nome, linewidth=2)

plt.axhline(0.35, color='orange', linestyle='--', linewidth=1, label='limiar observacao (0.35)')
plt.axhline(0.65, color='red', linestyle='--', linewidth=1, label='limiar critico (0.65)')
plt.xlabel('Tempo de missao (min)')
plt.ylabel('Estado do memristor (w)')
plt.title('Estado do memristor ao longo da missao em Marte')
plt.legend(loc='upper left', fontsize=8)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
